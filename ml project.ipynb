## train_test split

```python
df
```

```python
# Independent variables (Input features)
X = df.drop(['satisfaction_binary',"id"],axis=1)

# Dependent variable (Output/Target)
y = df['satisfaction_binary']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
categorical_columns = [
    "gender",
    "customer_type",
    "type_of_travel",
    "class"
]
```

```python
numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print(numerical_columns)
```

## feature Encoding

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_columns),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
```

```python
feature_names = preprocessor.get_feature_names_out()

print(feature_names)
```

```python
if hasattr(X_train_processed, "toarray"):
    processed_array = X_train_processed.toarray()
else:
    processed_array = X_train_processed
X_train_processed_df = pd.DataFrame(
    processed_array,
    columns=feature_names,
    index=X_train.index
)

X_train_processed_df
```

```python
print("X_train columns:")
print(X_train.columns.tolist())



print("\nCategorical columns:")
print(categorical_columns)
```

## Feature Scaling

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_columns),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ],
    remainder="drop"
)
```

```python
preprocessor.fit(X_train)

print("Preprocessor fitted successfully!")
```

```python
# Transform training features
X_train_transformed = preprocessor.transform(X_train)

# Transform testing features
X_test_transformed = preprocessor.transform(X_test)

print("Transformed training shape:", X_train_transformed.shape)
print("Transformed testing shape:", X_test_transformed.shape)
```

### Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import FunctionTransformer

nb_preprocessor = Pipeline([
    ('preprocessor', preprocessor),
    ('to_dense', FunctionTransformer(
        lambda x: x.toarray() if hasattr(x, 'toarray') else x
    ))
])

nb_model = Pipeline([
    ('preprocessor', nb_preprocessor),

    ('classifier', GaussianNB())
])
```

```python
nb_model.fit(X_train, y_train)

print("Naive Bayes Model Trained Successfully!")
```

```python
y_pred_nb = nb_model.predict(X_test)
```

### Overfitting & Underfitting Check 

```python
import pandas as pd

models = {
    'Logistic Regression': model,
    'KNN': knn_model,
    'Naive Bayes': nb_model,
    'SVC': svc_model,
    'Decision Tree': dtc_model
}

results = []

for name, model in models.items():

    train_accuracy = model.score(X_train, y_train)

    test_accuracy = model.score(X_test, y_test)

    gap = train_accuracy - test_accuracy

    results.append({
        'Algorithm': name,
        'Training Accuracy': round(train_accuracy * 100, 2),
        'Testing Accuracy': round(test_accuracy * 100, 2),
        'Performance Gap': round(gap * 100, 2)
    })

comparison = pd.DataFrame(results)

print(comparison)
```

```python
import matplotlib.pyplot as plt

comparison.set_index('Algorithm')[
    ['Training Accuracy', 'Testing Accuracy']
].plot(kind='bar', figsize=(10, 5))

plt.title('Overfitting and Underfitting Check')
plt.ylabel('Accuracy (%)')
plt.xlabel('ML Algorithms')
plt.xticks(rotation=30)
plt.ylim(0, 100)
plt.legend()
plt.tight_layout()
plt.show()
```

### Hyperparameter Tuning – Decision Tree

```python
### Hyperparameter Tuning – Decision Tree

from sklearn.model_selection import GridSearchCV

# Define parameter combinations
param_grid = {
    'classifier__max_depth': [5, 10, 15, 20, None],
    'classifier__min_samples_split': [2, 5, 10],
    'classifier__min_samples_leaf': [1, 2, 4],
    'classifier__criterion': ['gini', 'entropy']
}

# Grid Search
grid_search = GridSearchCV(
    estimator=dtc_model,
    param_grid=param_grid,
    cv=5,
    scoring='f1',
    n_jobs=-1,
    verbose=1
)

# Train Grid Search
grid_search.fit(X_train, y_train)

# Best Parameters
print("Best Parameters:")
print(grid_search.best_params_)

# Best CV Score
print("Best CV Score:")
print(grid_search.best_score_)
```

```python
from sklearn.metrics import accuracy_score, classification_report
from sklearn.metrics import confusion_matrix

# Best model
best_model = grid_search.best_estimator_

# Prediction
y_pred_tuned = best_model.predict(X_test)

# Evaluation
print("Tuned Model Accuracy:",
      accuracy_score(y_test, y_pred_tuned))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_tuned))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_tuned))
```

### Cross validation

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    best_model,
    X_train,
    y_train,
    cv=5,
    scoring='accuracy'
)

print("CV Scores:", scores)
print("Mean CV Accuracy:", scores.mean())
```

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import make_scorer, f1_score
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
models = {
    'Logistic Regression': model,
    'KNN': knn_model,
    'Naive Bayes': nb_model,
    'SVC': svc_model,
    'Decision Tree': dtc_model
}

cv_results = []

for name, estimator in models.items():

    scores = cross_val_score(
        estimator,
        X_train,
        y_train,
        cv=cv,
        scoring='accuracy',
        n_jobs=-1
    )

    cv_results.append({
        'Algorithm': name,
        'CV Mean Accuracy': scores.mean() * 100,
        'CV Standard Deviation': scores.std() * 100
    })

cv_df = pd.DataFrame(cv_results)

cv_df = cv_df.sort_values(
    by='CV Mean Accuracy',
    ascending=False
)

print(cv_df.round(2))
```
